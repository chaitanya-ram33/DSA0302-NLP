```

import re

text = "Python is a programming language. Python is easy to learn."

# Search for a word
result = re.search("Python", text)

if result:
    print("Pattern found:", result.group())
else:
    print("Pattern not found")

# Find all occurrences
matches = re.findall("Python", text)
print("All matches:", matches)

# Match a pattern at the beginning
match = re.match("Python", text)

if match:
    print("Match at beginning:", match.group())

# Replace a word
new_text = re.sub("Python", "Java", text)
print("After replacement:", new_text)
```
