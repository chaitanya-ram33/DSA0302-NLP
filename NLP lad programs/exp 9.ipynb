```

import re

# Step 1: Define POS tagging rules
patterns = [
    (r'.*ing$', 'VBG'),    # Present participle
    (r'.*ed$', 'VBD'),     # Past tense verb
    (r'.*ly$', 'RB'),      # Adverb
    (r'.*ful$', 'JJ'),     # Adjective
    (r'.*ness$', 'NN'),    # Noun
    (r'.*', 'NN')          # Default noun
]

# Step 2: Input words
words = [
    "playing",
    "walked",
    "quickly",
    "beautiful",
    "happiness",
    "book"
]

# Step 3: Apply POS tagging rules
print("Rule-Based POS Tagging Results:")

for word in words:
    for pattern, tag in patterns:

        if re.fullmatch(pattern, word.lower()):
            print(word, ":", tag)
            break
```
