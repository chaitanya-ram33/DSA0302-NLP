```

# EARLEY PARSER FOR CONTEXT-FREE GRAMMAR

# Step 1: Define Context-Free Grammar
grammar = {
    "S": [["NP", "VP"]],

    "NP": [
        ["Det", "N"],
        ["Det", "Adj", "N"],
        ["Name"]
    ],

    "VP": [
        ["V", "NP"],
        ["V"]
    ],

    "Det": [
        ["the"],
        ["a"],
        ["an"]
    ],

    "Adj": [
        ["big"],
        ["small"],
        ["beautiful"]
    ],

    "N": [
        ["boy"],
        ["girl"],
        ["dog"],
        ["cat"],
        ["apple"],
        ["book"]
    ],

    "V": [
        ["eats"],
        ["likes"],
        ["reads"],
        ["plays"]
    ],

    "Name": [
        ["John"],
        ["Mary"]
    ]
}


# Step 2: Define Earley Parser
def earley_parser(sentence):

    words = sentence.lower().split()
    n = len(words)

    # Create chart
    chart = [set() for _ in range(n + 1)]

    # Earley state format:
    # (LHS, RHS, dot_position, start_position)

    # Step 3: Initialize start state
    chart[0].add(("START", ("S",), 0, 0))

    # Step 4: Process each chart position
    for i in range(n + 1):

        changed = True

        while changed:

            changed = False

            for state in list(chart[i]):

                lhs, rhs, dot, start = state

                # Case 1: Predictor
                if dot < len(rhs):

                    symbol = rhs[dot]

                    if symbol in grammar:

                        for production in grammar[symbol]:

                            new_state = (
                                symbol,
                                tuple(production),
                                0,
                                i
                            )

                            if new_state not in chart[i]:

                                chart[i].add(new_state)
                                changed = True

                    # Case 2: Scanner
                    else:

                        if i < n and symbol.lower() == words[i]:

                            new_state = (
                                lhs,
                                rhs,
                                dot + 1,
                                start
                            )

                            chart[i + 1].add(new_state)

                # Case 3: Completer
                else:

                    for old_state in list(chart[start]):

                        old_lhs, old_rhs, old_dot, old_start = old_state

                        if old_dot < len(old_rhs):

                            if old_rhs[old_dot] == lhs:

                                new_state = (
                                    old_lhs,
                                    old_rhs,
                                    old_dot + 1,
                                    old_start
                                )

                                if new_state not in chart[i]:

                                    chart[i].add(new_state)
                                    changed = True

    # Step 5: Display chart states
    print("\nEARLEY PARSER CHART")

    for i in range(n + 1):

        print("\nChart", i)

        for state in chart[i]:

            lhs, rhs, dot, start = state

            production = list(rhs)
            production.insert(dot, "•")

            print(
                lhs,
                "->",
                " ".join(production),
                "[", start, ",", i, "]"
            )

    # Step 6: Check final state
    final_state = ("START", ("S",), 1, 0)

    if final_state in chart[n]:
        return True
    else:
        return False


# Step 7: Get input sentence
sentence = input("Enter a sentence: ")

print("\nInput Sentence:", sentence)

# Step 8: Execute parser
result = earley_parser(sentence)

# Step 9: Display result
print("\nParsing Result:")

if result:
    print("Sentence is ACCEPTED by the Grammar.")
else:
    print("Sentence is REJECTED by the Grammar.")
```
