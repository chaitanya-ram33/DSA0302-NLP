```

# TOP-DOWN PARSER FOR CONTEXT-FREE GRAMMAR

# Step 1: Define Context-Free Grammar
grammar = {
    "S": [["NP", "VP"]],

    "NP": [
        ["Det", "N"],
        ["Det", "Adj", "N"],
        ["Name"]
    ],

    "VP": [
        ["V", "NP"],
        ["V"]
    ],

    "Det": [
        ["the"],
        ["a"],
        ["an"]
    ],

    "Adj": [
        ["beautiful"],
        ["small"],
        ["big"]
    ],

    "N": [
        ["boy"],
        ["girl"],
        ["dog"],
        ["cat"],
        ["apple"],
        ["book"]
    ],

    "V": [
        ["eats"],
        ["likes"],
        ["reads"],
        ["plays"]
    ],

    "Name": [
        ["John"],
        ["Mary"]
    ]
}


# Step 2: Input sentence
sentence = input("Enter a sentence: ").strip()

# Convert sentence into words
words = sentence.split()

# Step 3: Initialize parsing counter
parse_count = 0


# Step 4: Define recursive top-down parser
def top_down_parse(symbols, position, tree):

    global parse_count

    # If all symbols are processed
    if not symbols:

        # Check whether all input words are matched
        if position == len(words):
            parse_count += 1
            print("\nParse Tree:")
            print(tree)
            return True

        return False

    # Get the first symbol
    current = symbols[0]

    # Remaining symbols
    remaining = symbols[1:]

    # Step 5: Check terminal symbol
    if current not in grammar:

        if position < len(words) and current.lower() == words[position].lower():

            return top_down_parse(
                remaining,
                position + 1,
                tree + " " + current
            )

        return False

    # Step 6: Expand non-terminal symbol
    found = False

    for production in grammar[current]:

        new_symbols = production + remaining

        result = top_down_parse(
            new_symbols,
            position,
            tree + " (" + current + " "
        )

        if result:
            found = True

    return found


# Step 7: Start parsing from S
print("\nInput Sentence:", sentence)
print("\nStarting Top-Down Parsing...")

result = top_down_parse(["S"], 0, "")


# Step 8: Display final result
print("\nTotal Parse Trees:", parse_count)

if result:
    print("\nSentence is ACCEPTED by the Grammar.")
else:
    print("\nSentence is REJECTED by the Grammar.")
```
