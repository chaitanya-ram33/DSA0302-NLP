```

import nltk

# Define CFG grammar for subject-verb agreement
grammar = nltk.CFG.fromstring("""
S -> SINGULAR_S | PLURAL_S

SINGULAR_S -> SINGULAR_NP SINGULAR_VP
PLURAL_S -> PLURAL_NP PLURAL_VP

SINGULAR_NP -> 'the' 'boy' | 'the' 'girl'
SINGULAR_NP -> 'a' 'boy' | 'a' 'girl'

PLURAL_NP -> 'the' 'boys' | 'the' 'girls'

SINGULAR_VP -> 'runs' | 'plays' | 'eats'
PLURAL_VP -> 'run' | 'play' | 'eat'
""")

# Create parser
parser = nltk.ChartParser(grammar)

# Get input sentence
sentence = input("Enter a sentence: ").lower()

# Split sentence into words
words = sentence.split()

# Check subject-verb agreement
try:
    trees = list(parser.parse(words))

    if trees:
        print("\nSubject-Verb Agreement is CORRECT.")
        print("\nParse Tree:")
        for tree in trees:
            tree.pretty_print()
    else:
        print("\nSubject-Verb Agreement is INCORRECT.")

except ValueError:
    print("\nInvalid sentence.")
```
