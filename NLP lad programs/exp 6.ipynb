```

import random

# Step 1: Input text
text = "I love NLP. I love Python. Python is easy to learn."

# Step 2: Convert text into lowercase words
words = text.lower().replace(".", "").split()

# Step 3: Create Bigram model
bigrams = {}

for i in range(len(words) - 1):
    current_word = words[i]
    next_word = words[i + 1]

    if current_word not in bigrams:
        bigrams[current_word] = []

    bigrams[current_word].append(next_word)

# Step 4: Display Bigram model
print("Bigram Model:")
for word, next_words in bigrams.items():
    print(word, "->", next_words)

# Step 5: Generate text
current_word = "i"
generated_text = [current_word]

for i in range(10):
    if current_word in bigrams:
        next_word = random.choice(bigrams[current_word])
        generated_text.append(next_word)
        current_word = next_word
    else:
        break

# Step 6: Display generated text
print("\nGenerated Text:")
print(" ".join(generated_text))
```
