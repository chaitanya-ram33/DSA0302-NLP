```

def plural(noun):
    state = 0

    # Check the ending of the noun
    if noun.endswith(("s", "sh", "ch", "x", "z")):
        state = 1
    elif noun.endswith("y") and len(noun) > 1:
        if noun[-2] not in "aeiou":
            state = 2
        else:
            state = 0
    else:
        state = 0

    # Generate plural form
    if state == 1:
        return noun + "es"
    elif state == 2:
        return noun[:-1] + "ies"
    else:
        return noun + "s"


# Test nouns
words = ["cat", "bus", "box", "baby", "toy", "dish"]

print("Plural Forms:")

for word in words:
    print(word, "->", plural(word))
```
