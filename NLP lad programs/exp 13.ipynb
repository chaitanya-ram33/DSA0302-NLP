```

# GENERATE PARSE TREE USING CONTEXT-FREE GRAMMAR

import nltk

# Step 1: Define Context-Free Grammar
grammar = nltk.CFG.fromstring("""
S -> NP VP

NP -> Det N
NP -> Det Adj N
NP -> Name

VP -> V NP
VP -> V

Det -> 'the' | 'a' | 'an'

Adj -> 'big' | 'small' | 'beautiful'

N -> 'boy' | 'girl' | 'dog' | 'cat' | 'apple' | 'book'

V -> 'eats' | 'likes' | 'reads' | 'plays'

Name -> 'john' | 'mary'
""")

# Step 2: Create a Chart Parser
parser = nltk.ChartParser(grammar)

# Step 3: Get input sentence
sentence = input("Enter a sentence: ").lower()

# Step 4: Tokenize the sentence
words = sentence.split()

# Step 5: Generate parse tree
print("\nInput Sentence:", sentence)
print("\nParse Tree:")

# Step 6: Parse and display the tree
trees = list(parser.parse(words))

if trees:

    for tree in trees:
        print(tree)
        print("\nTree Structure:")
        tree.pretty_print()

    print("\nSentence is ACCEPTED by the Grammar.")

else:
    print("No Parse Tree Found.")
    print("Sentence is REJECTED by the Grammar.")
```
