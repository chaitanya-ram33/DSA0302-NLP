```

import nltk

# Step 1: Training data
training_data = [
    [("The", "DT"), ("dog", "NN"), ("runs", "VBZ")],
    [("The", "DT"), ("cat", "NN"), ("eats", "VBZ")],
    [("A", "DT"), ("boy", "NN"), ("plays", "VBZ")]
]

# Step 2: Create initial tagging
words = ["The", "dog", "runs", "quickly"]

tagged_words = [(word, "NN") for word in words]

# Step 3: Define transformation rules
def transformation_tagging(tagged_words):

    result = []

    for i, (word, tag) in enumerate(tagged_words):

        # Rule 1: Articles are tagged as DT
        if word.lower() in ["the", "a", "an"]:
            tag = "DT"

        # Rule 2: Words ending in 'ly' are tagged as RB
        elif word.lower().endswith("ly"):
            tag = "RB"

        # Rule 3: Words ending in 's' after a noun are tagged as VBZ
        elif word.lower().endswith("s") and i > 0:
            if tagged_words[i-1][1] == "NN":
                tag = "VBZ"

        result.append((word, tag))

    return result

# Step 4: Apply transformation rules
final_tags = transformation_tagging(tagged_words)

# Step 5: Display results
print("Transformation-Based POS Tagging:")
for word, tag in final_tags:
    print(word, ":", tag)
```
