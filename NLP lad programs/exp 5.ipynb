```

from nltk.stem import PorterStemmer

# Create Porter Stemmer object
ps = PorterStemmer()

# List of words
words = [
    "playing",
    "played",
    "plays",
    "running",
    "runner",
    "connected",
    "connection",
    "studies"
]

print("Porter Stemming Results:")

for word in words:
    stem = ps.stem(word)
    print(word, "->", stem)
```
