```

import nltk

# Define Probabilistic CFG
grammar = nltk.PCFG.fromstring("""
S -> NP VP [1.0]
NP -> Det N [0.6] | 'John' [0.4]
VP -> V NP [0.7] | V [0.3]
Det -> 'the' [0.5] | 'a' [0.5]
N -> 'boy' [0.5] | 'apple' [0.5]
V -> 'eats' [0.6] | 'likes' [0.4]
""")

# Create Viterbi parser
parser = nltk.ViterbiParser(grammar)

# Input sentence
sentence = "the boy eats an apple"
sentence = sentence.replace("an", "a").lower().split()

# Parse sentence
print("Probabilistic Parse Tree:")

for tree in parser.parse(sentence):
    print(tree)
    print("Probability:", tree.prob())
    tree.pretty_print()
```
