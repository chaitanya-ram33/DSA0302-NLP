```
def vowel_groups(s):
    count = 0
    inside = False

    for c in s:
        if c in "aeiou":
            if not inside:
                count += 1
            inside = True
        else:
            inside = False

    return count


def doublable(s):
    return (
        len(s) >= 3
        and s[-3] not in "aeiou"
        and s[-2] in "aeiou"
        and s[-1] not in "aeiouwxy"
        and vowel_groups(s) == 1
    )


def inflect(s, f):
    if f in ("N+PL", "V+3SG"):
        if s.endswith(("s", "x", "z", "ch", "sh")):
            return s + "es"

        if len(s) >= 2 and s.endswith("o") and s[-2] not in "aeiou":
            return s + "es"

        if len(s) >= 2 and s.endswith("y") and s[-2] not in "aeiou":
            return s[:-1] + "ies"

        return s + "s"

    if f == "V+PROG":
        if len(s) >= 3 and s.endswith("e") and s[-2] not in "aeiou":
            return s[:-1] + "ing"

        if doublable(s):
            return s + s[-1] + "ing"

        return s + "ing"

    suffix = {
        "V+PAST": "ed",
        "ADJ+COMP": "er",
        "ADJ+SUP": "est"
    }[f]

    if s.endswith("e"):
        if suffix == "ed":
            return s + "d"
        if suffix == "er":
            return s + "r"
        return s + "st"

    if len(s) >= 2 and s.endswith("y") and s[-2] not in "aeiou":
        return s[:-1] + "i" + suffix

    if doublable(s):
        return s + s[-1] + suffix

    return s + suffix


S = int(input())
lexicon = []

for _ in range(S):
    stem, pos = input().split()
    lexicon.append((stem, pos))

Q = int(input())

for _ in range(Q):
    word = input().strip()
    result = []

    for stem, pos in lexicon:
        if word == stem:
            result.append(stem + "+" + pos)

        if pos == "N":
            features = [("N+PL", "PL")]

        elif pos == "V":
            features = [
                ("V+3SG", "3SG"),
                ("V+PROG", "PROG"),
                ("V+PAST", "PAST")
            ]

        else:
            features = [
                ("ADJ+COMP", "COMP"),
                ("ADJ+SUP", "SUP")
            ]

        for feature, label in features:
            if inflect(stem, feature) == word:
                result.append(stem + "+" + pos + "+" + label)

    result = sorted(set(result))

    if result:
        print(", ".join(result))
    else:
        print("UNKNOWN")
```

```
def consonant(w, i):
    c = w[i]

    if c in "aeiou":
        return False

    if c == "y" and i > 0:
        return not consonant(w, i - 1)

    return True


def measure(w):
    m = 0

    for i in range(1, len(w)):
        if not consonant(w, i - 1) and consonant(w, i):
            m += 1

    return m


def has_vowel(w):
    for i in range(len(w)):
        if not consonant(w, i):
            return True

    return False


def double_consonant(w):
    return (
        len(w) >= 2
        and w[-1] == w[-2]
        and consonant(w, len(w) - 1)
    )


def cvc(w):
    if len(w) < 3:
        return False

    n = len(w)

    return (
        consonant(w, n - 3)
        and not consonant(w, n - 2)
        and consonant(w, n - 1)
        and w[-1] not in "wxy"
    )


def stem(w):
    # Step 1a
    if w.endswith("sses"):
        w = w[:-2]
    elif w.endswith("ies"):
        w = w[:-2]
    elif w.endswith("ss"):
        pass
    elif w.endswith("s"):
        w = w[:-1]

    # Step 1b
    if w.endswith("eed"):
        if measure(w[:-3]) > 0:
            w = w[:-1]

    elif w.endswith("ed") and has_vowel(w[:-2]):
        w = w[:-2]

        if w.endswith(("at", "bl", "iz")):
            w += "e"
        elif double_consonant(w) and w[-1] not in "lsz":
            w = w[:-1]
        elif measure(w) == 1 and cvc(w):
            w += "e"

    elif w.endswith("ing") and has_vowel(w[:-3]):
        w = w[:-3]

        if w.endswith(("at", "bl", "iz")):
            w += "e"
        elif double_consonant(w) and w[-1] not in "lsz":
            w = w[:-1]
        elif measure(w) == 1 and cvc(w):
            w += "e"

    # Step 1c
    if w.endswith("y") and has_vowel(w[:-1]):
        w = w[:-1] + "i"

    return w


Q = int(input())

for _ in range(Q):
    word = input().strip()
    print(stem(word))
```
